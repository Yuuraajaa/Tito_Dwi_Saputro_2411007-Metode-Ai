```python
import os
import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO
from IPython.display import display
```

```python
model = YOLO("yolov8n.pt")

print("Model YOLOv8 berhasil dimuat!")
```

```python
image_paths = [
    "/content/drive/MyDrive/5 gambar/bangyu-wang-4Y4HCX3fYtw-unsplash.jpg",
    "/content/drive/MyDrive/5 gambar/jason-leung-jo0e5a0x2A8-unsplash.jpg",
    "//content/drive/MyDrive/5 gambar/joaquin-arenas-gBTjTvqan7g-unsplash.jpg",
    "//content/drive/MyDrive/5 gambar/ronaldo-liu-TaFBn_uK2HA-unsplash.jpg",
    "//content/drive/MyDrive/5 gambar/tony-hsu-o1gMxVgnhWY-unsplash.jpg"
]
```

```python
os.makedirs("hasil_deteksi", exist_ok=True)

for i, image_path in enumerate(image_paths, start=1):
    print(f"\nMemproses gambar {i}: {image_path}")

    if not os.path.exists(image_path):
        print("Gambar tidak ditemukan, periksa kembali path!")
        continue

    results = model.predict(
        source=image_path,
        conf=0.25,
        imgsz=640,
        verbose=False
    )

    result = results[0]

    image_result = result.plot()

    output_path = f"hasil_deteksi/hasil_{i}.jpg"
    cv2.imwrite(output_path, image_result)

    image_rgb = cv2.cvtColor(image_result, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(10, 7))
    plt.imshow(image_rgb)
    plt.title(f"Hasil Deteksi YOLOv8 - Gambar {i}")
    plt.axis("off")
    plt.show()

    print("Objek yang terdeteksi:")

    if len(result.boxes) > 0:
        for box in result.boxes:
            class_id = int(box.cls[0])
            confidence = float(box.conf[0])
            object_name = model.names[class_id]

            print(
                f"- {object_name}: "
                f"Confidence {confidence:.2%}"
            )
    else:
        print("Tidak ada objek yang terdeteksi.")

    print(f"Hasil disimpan di: {output_path}")

print("\nProses deteksi kelima gambar selesai!")
```
